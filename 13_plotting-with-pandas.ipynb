# Plotting with Pandas

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("https://github.com/ami-null/python-course/raw/refs/heads/main/data/Top_10000_Movies_IMDb.csv")
```

```python
df['genre_one'] = df['Genre'].str.split(',').str[0]
```

```python
df['genre_one'].value_counts()
```

## Bar diagram

```python
df.genre_one.value_counts().plot(
    kind = "bar",
    title = "Number of movies by year",
    xlabel = "Genre",
    ylabel = "Number of movies"
)
```

`normalize=True` provides relative frequency:

```python
df.genre_one.value_counts(normalize=True).plot(
    kind = "bar",
    title = "Number of movies by year",
    xlabel = "Genre",
    ylabel = "Number of movies"
)
```

## Histogram

```python
# histogram
df.Rating.plot(
    kind = "hist",
    bins = 20,
    color = "skyblue",
    edgecolor = "white",
    density = True,
    # ylabel = "Relative frequency",
    # xlabel="Rating"
)
# THE ABOVE TWO ARGUMENTS DO NOT WORK IN OLDER VERSIONS OF PANDAS
# INSTEAD USE:
plt.ylabel("Relative frequency")
plt.xlabel("Rating")
```

## Boxplot

```python
df.Rating.plot(kind="box")
```

### Grouped boxplots

```python
df.boxplot(
    column = "Rating",
    by = "genre_one",
    grid = False
)
plt.xticks(rotation=90);
```

## Scatter plot

```python
df.plot(
    kind = "scatter",
    x = "Rating",
    y = "Gross"
)

plt.savefig("scatter.png")
```

## Pie chart

```python
df.genre_one.value_counts().plot(kind = "pie")
```

Removing genres with fewer than 50 movies in the dataset:

```python
df2 = df.groupby('genre_one').filter(lambda x: len(x) > 50)
```

```python
df2.genre_one.value_counts().plot(kind = "pie")
```

## Density plot

```python
df.Rating.plot(kind = "density")
```
