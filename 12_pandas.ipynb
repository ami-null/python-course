# Pandas

`pandas` is an open-source module for Python that contains data structures and data manipulation tools designed to make data cleaning and analysis fast and convenient in Python.

It adopts significant parts of NumPy's idiomatic style of array-based computing, especially array-based functions and a preference for data processing without for loops.

While `pandas` adopts many coding idioms from `NumPy`, the biggest difference is that `pandas` is designed for working with tabular or heterogeneous data. `NumPy`, by contrast, is best suited for working with homogeneously typed numerical array of one or more dimensions.

### A common pandas pitfall: chained indexing

When you filter a `DataFrame` and then try to assign a new value to the filtered result in the same statement, pandas cannot always tell whether you are working with a view or a copy of the original data. This produces a `SettingWithCopyWarning`, and the assignment may silently fail to update anything.

The fix is to explicitly assign the filtered result to a new variable with `.copy()`, so pandas knows you intend to work with an independent copy rather than a view of the original data.

````python
# UNCOMMENT TO SEE THE WARNING
# chained indexing: filter, then assign, in two separate bracket operations, without copying first
# high_rated = movies_df[movies_df['Rating'] > 9]
# high_rated['Genre'] = 'Top Tier'    # triggers SettingWithCopyWarning
````

````python
# correct approach: .copy() makes the new variable an independent DataFrame, not a view
high_rated = movies_df[movies_df['Rating'] > 9].copy()
high_rated['Genre'] = 'Top Tier'
high_rated.head()
````

### Discretization and Binning

Sometimes we need to convert numeric/continuous columns to categorical/discrete columns.

#### Using `pd.cut()`

This functions "cuts" a numeric/continuous variable into several groups making it discrete or categorical.

```
pandas.cut(x, bins, right=True, labels=None, retbins=False, precision=3, include_lowest=False, duplicates='raise', ordered=True)
```

Parameters:
- x: array-like (series/column). The input array to be binned. Must be 1-dimensional.

- bins: int, sequence of scalars, or IntervalIndex. The criteria to bin by.
    - int : Defines the number of equal-width bins in the range of x. The range of x is extended by .1% on each side to include the minimum and maximum values of x.
    - sequence of scalars : Defines the bin edges allowing for non-uniform width. No extension of the range of x is done.
    - IntervalIndex : Defines the exact bins to be used. Note that IntervalIndex for bins must be non-overlapping.
- right: bool, default=True. Indicates whether bins includes the rightmost edge or not. If right == True (the default), then the bins [1, 2, 3, 4] indicate (1,2], (2,3], (3,4]. This argument is ignored when bins is an IntervalIndex.
- labels: array or False, default=None. Specifies the labels for the returned bins. Must be the same length as the resulting bins.
- include_lowest: bool, default=False. Whether the first interval should be left-inclusive or not.
- ordered: bool, default=True. Whether the labels are ordered or not. Applies to returned types Categorical and Series (with Categorical dtype). If True, the resulting categorical will be ordered. If False, the resulting categorical will be unordered (labels must be provided).



````python
rating_cat = pd.cut(movies_df['Rating'], 3)
rating_cat.value_counts()
````

Can mention `right = False` to include lower limits and not upper limits:

````python
rating_cat2 = pd.cut(movies_df['Rating'], 3, right = False)
rating_cat2.value_counts()
````

````python
rating_cat3 = pd.cut(movies_df['Rating'], 3, include_lowest=True)
rating_cat3.value_counts()
````

````python
rating_cat4 = pd.cut(movies_df.Rating, [0, 5, 8, 10], right = True)
rating_cat4.value_counts()
````

Can also provide a `list` of labels for each category using the `labels` argument:

````python
rating_cat5 = pd.cut(movies_df.Rating, [0, 5, 8, 10], labels = ['<=5', '5-8', '>8'])
rating_cat5.value_counts()
````

#### `pd.qcut()`

Quantile-based discretization function. Uses sample quantiles to bin.

```
Parameters
----------
x : 1d ndarray or Series
q : int or list-like of float
    int: Number of quantiles. 10 for deciles, 4 for quartiles, etc.
    list: Alternately array of quantiles, e.g. [0, .25, .5, .75, 1.] for quartiles.
labels : array or False, default None
    Used as labels for the resulting bins. Must be of the same length as
    the resulting bins. If False, return only integer indicators of the
    bins. If True, raises an error.
retbins : bool, optional
    Whether to return the (bins, labels) or not. Can be useful if bins
    is given as a scalar.
```

````python
pd.qcut(x = movies_df.Rating, q = 3).value_counts()
````

### Applying Functions and Mapping


A frequent operation while analyzing a data is to apply a function on one-dimensional arrays to each column or each row. It can be done with the `.apply()` method of a `DataFrame`.

The `.apply()` method takes a function as its argument as well as the axis *across* which the function is applied. The function can be built-in, from a package or a lambda.

For example, we can define a function that returns the range and the 90th percentile of a variable:

````python
def mysummary(x):
    return [x.max() - x.min(), x.quantile(0.90)]
````

If we want to find the range and the median of say, rating and gross revenue:

````python

movies_df[['Rating', 'Gross']].apply(mysummary, axis=0)
````

We can see that the two summary statistics returned from our defined function, namely, range and 90th percentile are indexed as 0 and 1, respectively in the output.

While outputting multiple summary statistics, it is desirable to index those statistics with appropriate labels.

We can provide appropriate labels to our outputs by returning a Series instead of a list as follows:

````python
def mysummary(x):
    return pd.Series([x.max() - x.min(), x.quantile(0.90)],
                    index = ['range', '90th Percentile'])
movies_df[['Runtime', 'Gross']].apply(mysummary, axis=0)
````

We can also invoke lambda functions inside the `.apply()` method instead of defining a function using the `def` keyword:

````python
movies_df[['Runtime', 'Gross']].apply(
    lambda x: pd.Series(
        [x.max() - x.min(), x.quantile(0.90)],
        index = ['range', '90th Percentile']
    )
)
````

Using apply() will be much faster than iterating manually over rows because pandas is utilizing vectorization.

### More `.groupby` examples

**Note on the dataset used below**

The examples in this subsection switch away from `movies_df` to a small, hand built state population dataset. This is intentional: the point here is to isolate the mechanics of `.groupby()` with weighted averages, custom aggregation via `.apply()`, and `.transform()`, without the noise of a large real dataset. We return to `movies_df` in the next section on joining dataframes.

#### Groupby the summarize/aggregate

````python
data = {
    "state" : ["Ohio"]*3 + ["Nevada"]*3,
    "year" : [2000, 2001, 2002, 2001, 2002, 2003],
    "popn" : [1.5, 1.7, 3.6, 2.4, 2.9, 3.2],
    "wt" : [1, 2, 3] * 2    # weights for findingg weighted average
}
df = pd.DataFrame(data)
df
````

````python
df.groupby("state").agg(avg_pop = ("popn", "mean"))    # returns a dataframe
````

````python
# finding weighted average
df.groupby("state").apply(lambda g: sum(g.popn * g.wt) / sum(g.wt))    # returns a Series
````

````python
# wrapping the result in a Series makes the whole thing a dataframe
df.groupby("state").apply(
    lambda g: pd.Series(
        {"wt_avg" : sum(g.popn * g.wt) / sum(g.wt)}
    )
)
````

````python
# this returns a dataframe
agdf = (
    df.groupby(df.state).apply(
        lambda x: pd.Series(
            {
                "avg" : np.mean(x.popn),
                "wt_avg" : sum(x.popn * x.wt)/sum(x.wt),
                "wt_sum" : sum(x.popn * x.wt)
            }
            )
        )
    )

agdf
````

#### Groupby then mutate

````python
df.join(agdf, on = "state")
````

````python
df.groupby(df.state)["popn"].transform(lambda x: x.mean())
````

````python
df["avg_pop2"] = df.groupby(df.state)["popn"].transform(lambda x: x.mean())
df
````

````python
(
    df
    .groupby("state", group_keys = False)
    .apply(
        lambda grp: grp.assign(
            wt_avg2 = lambda x: sum(x.popn * x.wt) / sum(x.wt)
        ),
    )
)
````

## Joining dataframes

`pd.merge()` (or the equivalent `.merge()` method) combines two `DataFrame`s based on the values in one or more shared key columns, similar to a SQL join.

Key parameters:

- `on`: column name (or list of names) to join on, when both frames use the same column name for the key.
- `left_on` / `right_on`: used instead of `on` when the key column has different names in each frame.
- `how`: the join type.
    - `inner` (default): keep only keys present in both frames.
    - `left`: keep all keys from the left frame, filling missing right-side values with `NaN`.
    - `right`: keep all keys from the right frame, filling missing left-side values with `NaN`.
    - `outer`: keep all keys from either frame, filling missing values with `NaN` on whichever side lacks the key.
- `suffixes`: a tuple of strings appended to overlapping non-key column names, so both can be kept without a name clash.
- `indicator`: if `True`, adds a `_merge` column labelling each row as `left_only`, `right_only`, or `both`.

````python
df1 = movies_df[['ID', 'title', 'Runtime']]
df2 = movies_df[['ID', 'title', 'Rating']]
````

````python
dfjoin = df1.merge(df2, on = "ID", how = "inner")
dfjoin.info()
````
